```python
import cv2
import os

# -----------------------------
# Person name
# -----------------------------
PERSON_NAME = input("Enter person's name: ").strip()

# Create folder for this person
save_dir = os.path.join("dataset", PERSON_NAME)
os.makedirs(save_dir, exist_ok=True)

# -----------------------------
# Face detector
# -----------------------------
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades +
    "haarcascade_frontalface_default.xml"
)

# -----------------------------
# Webcam
# -----------------------------
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("❌ Cannot open webcam")
    exit()

# Count existing images
count = len([
    f for f in os.listdir(save_dir)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
])

print("\n================================")
print("📸 Face Dataset Collection")
print("================================")
print(f"Person: {PERSON_NAME}")
print(f"Folder: {save_dir}")
print("\nPress 'S' to save a photo")
print("Press 'Q' to quit")
print("================================\n")


while True:

    ret, frame = cap.read()

    if not ret:
        print("❌ Failed to read camera")
        break

    # Mirror camera
    frame = cv2.flip(frame, 1)

    # Convert to grayscale
    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    # Detect faces
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.1,
        minNeighbors=5,
        minSize=(100, 100)
    )

    # Draw face rectangles
    for (x, y, w, h) in faces:

        cv2.rectangle(
            frame,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

    # Information
    cv2.putText(
        frame,
        f"Person: {PERSON_NAME}",
        (20, 35),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 0),
        2
    )

    cv2.putText(
        frame,
        f"Photos: {count}",
        (20, 70),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.8,
        (0, 255, 0),
        2
    )

    cv2.putText(
        frame,
        "Press S = Save | Q = Quit",
        (20, 110),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (255, 255, 255),
        2
    )

    cv2.imshow(
        "Face Dataset Collection",
        frame
    )

    # Keyboard
    key = cv2.waitKey(1) & 0xFF

    # -----------------------------
    # Save photo with S
    # -----------------------------
    if key == ord("s"):

        if len(faces) == 0:

            print("⚠️ No face detected!")

        else:

            # If multiple faces exist,
            # save the largest face
            face = max(
                faces,
                key=lambda rect: rect[2] * rect[3]
            )

            x, y, w, h = face

            face_crop = frame[y:y+h, x:x+w]

            count += 1

            filename = os.path.join(
                save_dir,
                f"{PERSON_NAME}_{count}.jpg"
            )

            cv2.imwrite(
                filename,
                face_crop
            )

            print(f"✅ Saved: {filename}")

    # -----------------------------
    # Quit with Q
    # -----------------------------
    elif key == ord("q"):
        break


# -----------------------------
# Release
# -----------------------------
cap.release()
cv2.destroyAllWindows()

print("\n✅ Dataset collection finished.")
print(f"Total photos: {count}")
```
